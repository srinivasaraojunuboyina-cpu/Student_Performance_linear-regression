```python
x = df.drop(['Performance Index'], axis=1)
y = df['Performance Index']
```

```python
x
```

```python


df['Extracurricular Activities'] = pd.get_dummies(
    df['Extracurricular Activities'],
    drop_first=True,
    dtype=int
)
```

```python
y
```

```python
x
```

```python
 from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from  sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from  sklearn.linear_model import LinearRegression , Lasso , Ridge
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
y_predict = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error , mean_absolute_error,root_mean_squared_error,
r2_score(y_test, y_predict)
mean_absolute_error(y_test, y_predict)
mean_squared_error(y_test, y_predict)
root_mean_squared_error(y_test, y_predict)
```

```python
print(r2_score(y_test, y_predict))
print(mean_absolute_error(y_test, y_predict))
print(mean_squared_error(y_test, y_predict))
print(root_mean_squared_error(y_test, y_predict))
```

```python
lass = Lasso(alpha=0.5)
lass.fit(x_train, y_train)
```

```python
la_predict = lass.predict(x_test)
```

```python
r2_score(y_test, la_predict)
mean_absolute_error(y_test, la_predict)
mean_squared_error(y_test, la_predict)
root_mean_squared_error(y_test, la_predict)
```

```python
print(r2_score(y_test, la_predict))
print(mean_absolute_error(y_test, la_predict))
print(mean_squared_error(y_test, la_predict ))
print(root_mean_squared_error(y_test, la_predict))
```

```python
rid = Ridge(alpha=0.5)
rid.fit(x_train, y_train)
```

```python
rid_predict = rid.predict(x_test)
```

```python
r2_score(y_test, rid_predict)
mean_absolute_error(y_test, rid_predict)
mean_squared_error(y_test, rid_predict)
root_mean_squared_error(y_test, rid_predict)
```

```python
print(r2_score(y_test, rid_predict))
print(mean_absolute_error(y_test, rid_predict))
print(mean_squared_error(y_test, rid_predict))
print(root_mean_squared_error(y_test, rid_predict))
```

```python
#comparision of result
```

```python

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error
)

models = {
    "Linear Regression": y_predict,
    "Lasso Regression": la_predict,
    "Ridge Regression": rid_predict
}

for model_name, prediction in models.items():
    print(f"\n{model_name}")
    print("R2 Score:", r2_score(y_test, prediction))
    print("MAE:", mean_absolute_error(y_test, prediction))
    print("MSE:", mean_squared_error(y_test, prediction))
    print("RMSE:", root_mean_squared_error(y_test, prediction))
```

Linear Regression — Best overall (by a very small margin)

1. Linear Regression: Highest R² score (98.843%) and lowest MSE/RMSE.
2. Ridge Regression: Almost identical to Linear Regression, with a slightly lower MAE.
3. Lasso Regression: Lowest R² score and highest prediction errors.

"After comparing Linear, Lasso, and Ridge Regression using R², MAE, MSE, and RMSE, I found that Linear Regression performed slightly better overall, with an R² score of 98.84% and RMSE of 2.075. Ridge Regression gave almost identical results, while Lasso showed slightly higher errors."
Final conclusion: Linear Regression and Ridge Regression perform practically the same on this test set. Linear Regression is the simplest choice based on these metrics, but cross-validation would help confirm the best model.
